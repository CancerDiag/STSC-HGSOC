```python
data_folder = "./../data/st_data/"
sample_lst = os.listdir(data_folder)
chemo_resp = { "P1": 'poor', "P2": 'good', "P3": "good", "P4": "partial", "P5": "good", "P6": "partial", "P7": "poor", "P8": "poor"}
for sample in sample_lst:
    #print(idx, sample)

    patient = sample[12:14]
    adata = sc.read_visium( os.path.join(data_folder, sample), library_id = patient)
    adata.var_names_make_unique()
    adata.obs["sample"] = patient
    adata.obs["chemo_response"] = chemo_resp[patient]
    print(adata)
    
    adata.write_h5ad("./../data/st_raw/{}.h5ad".format(sample))
```

```python
adatas = dict()
for sample in sample_lst:
    adatas[sample] = sc.read_h5ad("./../data/st_raw/{}.h5ad".format(sample))
adatas
```

```python
## filtering the data
for sample in sample_lst:
    sc.pp.filter_genes(adatas[sample], min_cells=5)
    sc.pp.filter_cells(adatas[sample], min_counts=500)
adatas
```

```python
for sample in sample_lst:
    adatas[sample].write("./../data/st_qc/{}.h5ad".format(sample))
```
