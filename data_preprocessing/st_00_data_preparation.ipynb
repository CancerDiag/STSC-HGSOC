```r
library(DropletUtils)
library(Seurat)
library(stringr)
```

```r
files <- list.files("./../data/", pattern = "^GSM")
files
```

```r
inputDir <- "./../data/Read10x_data/"
outputDir <- "./../data/st_data/"

if (!dir.exists(file.path(inputDir))) {
      dir.create(file.path(inputDir))
    }

if (!dir.exists(file.path(outputDir))) {
      dir.create(file.path(outputDir))
    }
```

```r
for ( id in unique(substr(files, 1, 14))){
    #print(id)
    dirn <- paste(inputDir, id, sep = "")
    if (!dir.exists(file.path(dirn))) {
      dir.create(file.path(dirn))
    }
    
    id_files = list.files(inputDir, pattern = paste(id, ".", sep = ""))
    if (length(id_files)> 0){
        for (i in id_files){
            from <- paste(inputDir, i , sep = "")
            to <- paste( dirn, "/", i, sep = "")
            file.rename(from = from, to = to)
        }
    }
}
```

```r
for ( id in unique(substr(files, 1, 14))){
    filter_matrix <- Read10X(paste(inputDir, id , sep = ""))
    
    dirn <- paste(outputDir, id, sep = "")
    if (!dir.exists(file.path(dirn))) {
      dir.create(file.path(dirn))
    }
    
    write10xCounts(file.path(paste(dirn, "/filtered_feature_bc_matrix.h5", sep = "")), filter_matrix, type = "HDF5",
                   genome = "mm10", version = "3", overwrite = TRUE,
                   gene.id = rownames(filter_matrix),
                   gene.symbol = rownames(filter_matrix))

}
```

```r
for ( id in unique(substr(files, 1, 14))){
    fn = list.files("./../data/", pattern = id)
    from = paste("./../data/", fn, sep = "")
    to = paste(outputDir, id, "/", fn, sep = "")
    file.rename(from = from, to = to)
        
}
```
